```python
%load_ext autoreload
%autoreload 2
%matplotlib inline

import numpy as np
import math
import scipy
import copy
import os
import matplotlib.pyplot as plt

from htdeblur import blurkernel
from htdeblur import simulation
from htdeblur import linops
from htdeblur.roi import Roi
from htdeblur._util import pad
import nbconfig

nbconfig.use_style()
```

```python
figure_output_directory = nbconfig.FIGURE_DIR + '/'
```

## Load Object

```python
california = simulation.california(min_value=0, max_value=1, invert=False, shape=(320, 1000))
plt.figure()
plt.imshow(california)
plt.colorbar()
```

## Generate Example Blurred Images

```python
object_shape = (360, 1200)
image_shape = (360, 600)

# Generate object
object_true = np.abs(pad(california, object_shape, center=True))

# Define Crop ROIs
roi_list = (Roi(start=(0,0), shape=image_shape, input_shape=object_shape),
            Roi(start=(0, object_shape[1] // 4), shape=image_shape, input_shape=object_shape), 
            Roi(start=(0, object_shape[1] // 2), shape=image_shape, input_shape=object_shape))

# Generater Blur kernel
vector, dnf = blurkernel.vector(pulse_count=50)
kernel = blurkernel.fromVector(vector, object_shape)

# Define Operators
C = linops.Convolution(kernel)
G = linops.Segmentation(roi_list)

# Crop to measurements
measurement_list = linops.vector_split(G * C * object_true, 3)
```

## Save Ground Truth

```python
plt.figure()
plt.imshow(object_true)
plt.axis('off')
plt.savefig(figure_output_directory + 'system_ground_truth.pdf')
```

## Save Measurements

```python
plt.figure(figsize=(3,5))
plt.subplot(311)
plt.imshow(measurement_list[0])
plt.axis('off')
plt.subplot(312)
plt.imshow(measurement_list[1])
plt.axis('off')
plt.subplot(313)
plt.imshow(measurement_list[2])
plt.axis('off')
plt.tight_layout()
plt.savefig(figure_output_directory + 'system_measurements.pdf')
```

```python
plt.figure(figsize=(4,6))
plt.imshow(linops.vector_stack(measurement_list))
plt.axis('off')
plt.tight_layout()
plt.savefig(figure_output_directory + 'system_measurements_stack.pdf')
```

## Save Shifted Blur Kernels

```python
roi_crop = roi_list[1].decimate(6)
blur_kernel_zoom = blurkernel.fromVector(vector[:np.shape(vector)[0]//3], roi_crop.input_shape)

plt.figure(figsize=(4,5))
plt.subplot(311)
plt.imshow(np.roll(blur_kernel_zoom, (0, 75), axis=(0, 1)))
plt.axis('off')
plt.subplot(312)
plt.imshow(blur_kernel_zoom)
plt.axis('off')
plt.subplot(313)
plt.imshow(np.roll(blur_kernel_zoom, (0, -75), axis=(0, 1)))
plt.axis('off')
plt.tight_layout()
plt.savefig(figure_output_directory + 'system_kernel_stack.pdf')
```
